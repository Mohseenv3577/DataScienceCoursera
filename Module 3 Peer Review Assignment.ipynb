#Relationship Between Descriptive Statistics and the Box Plot#
The square diagram for the quantity visualizes the same data shown in the descriptive statistics.
The minimum value matches the lower whisker of the box plot.
The 25% is shown in the lower edge of the box.
The 50% is shown as the line inside the box.
The 75% is shown the upper edge of the box.
The maximum value is shown in the top level of the square diagram
